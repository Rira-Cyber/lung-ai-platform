```python
import torch

from backend.models.blocks import DownBlock

x = torch.randn(
    2,
    32,
    128,
    128,
    128,
)

block = DownBlock(
    in_channels=32,
    out_channels=64,
)

pooled, skip = block(x)

print(skip.shape)
print(pooled.shape)
```

```python
import torch

from backend.models.blocks import UpBlock

x = torch.randn(
    2,
    512,
    8,
    8,
    8,
)

skip = torch.randn(
    2,
    256,
    16,
    16,
    16,
)

block = UpBlock(
    in_channels=512,
    skip_channels=256,
    out_channels=256,
)

y = block(
    x,
    skip,
)

print(y.shape)
```
