```python
import torch
import setup
from backend.configs import DEV_CONFIG

from backend.training.dataloader import (
    create_train_loader,
)

from backend.models.unet3d import UNet3D

from backend.losses.dice import DiceLoss
from backend.metrics.dice import DiceMetric

from backend.training.trainer import Trainer
```

```python
config = DEV_CONFIG

device = torch.device("cpu")
```

```python
train_loader = create_train_loader(
    dataset_path=config.dataset_path,
    batch_size=config.batch_size,
    num_workers=config.num_workers,
    patch_size=config.patch_size,
    positive_ratio=config.positive_ratio,
)
```

```python
model = UNet3D(
    in_channels=1,
    out_channels=1,
    features=config.features,
)
```

```python
optimizer = torch.optim.Adam(
    model.parameters(),
    lr=config.learning_rate,
)
```

```python
criterion = DiceLoss()

metric = DiceMetric()
```

```python
trainer = Trainer(
    model=model,
    optimizer=optimizer,
    criterion=criterion,
    metric=metric,
    device=device,
)
```

```python
trainer.fit(
    train_loader=train_loader,
    epochs=1,
)
```
