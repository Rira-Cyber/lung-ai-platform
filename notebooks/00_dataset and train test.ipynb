```python
import setup

import torch

from backend.config import DEV_CONFIG

from backend.models.unet3d import UNet3D

from backend.preprocessing.patch_sampler import PatchSampler

from backend.training.dataloader import create_train_loader

from backend.training.trainer import Trainer

from backend.losses.dice import DiceLoss

from backend.metrics.dice import DiceMetric

```

```python

config = DEV_CONFIG

device = torch.device(config.device)

```

```python

train_loader = create_train_loader(
    dataset_path=config.dataset_path,
    batch_size=config.batch_size,
    num_workers=config.num_workers,
    patch_size=config.patch_size,
    positive_ratio=config.positive_ratio,
)
```

```python
model = UNet3D(
    in_channels=config.in_channels,
    out_channels=config.out_channels,
    features=config.features,
).to(device)


criterion = DiceLoss()

metric = DiceMetric()


optimizer = torch.optim.Adam(
    model.parameters(),
    lr=config.learning_rate,
)

```

```python

trainer = Trainer(
    model=model,
    optimizer=optimizer,
    criterion=criterion,
    metric=metric,
    device=device,
)


loss, dice = trainer.train_epoch(train_loader)

print(loss)
print(dice)
```
