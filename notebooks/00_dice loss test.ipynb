```python
import setup

import torch

from backend.models.unet3d import UNet3D

from backend.losses.dice import DiceLoss

from backend.metrics.dice import DiceMetric

from backend.training.trainer import Trainer

from backend.training.dataloader import create_train_loader
```

```python
device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

print(device)
```

```python
dataset_path = (
    r"E:\My codes\Medical AI\lung-ai-platform\data\raw\lidc_idri"
)
```

```python
train_loader = create_train_loader(
    dataset_path=dataset_path,
    batch_size=2,
    shuffle=True,
)
```

```python
model = UNet3D().to(device)

criterion = DiceLoss()

metric = DiceMetric()

optimizer = torch.optim.Adam(
    model.parameters(),
    lr=1e-3,
)
```

```python
trainer = Trainer(
    model=model,
    optimizer=optimizer,
    criterion=criterion,
    metric=metric,
    device=device,
)
```

```python
batch = next(iter(train_loader))

print(batch["image"].shape)

print(batch["mask"].shape)

print(batch["mask"].sum())
```

```python
batch = next(iter(train_loader))

loss, dice = trainer.train_step(batch)

print(loss)

print(dice)
```
